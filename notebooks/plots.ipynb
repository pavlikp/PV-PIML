```python
import numpy as np

df = pd.read_csv(r"C:\Users\peter\Downloads\wandb_export_2026-07-10T22_42_23.715+02_00.csv")

TSMixer = []
ADF_augment = []
ADF_residual = []

for i in range(10):
    TSMixer.append(df[f"TSMixer{i} - val_loss"])
    ADF_augment.append(df[f"TSMixerAugment{i} - val_loss"])
    ADF_residual.append(df[f"Residual_{i} - val_loss"])

TSMixer = np.vstack(TSMixer)
ADF_augment = np.vstack(ADF_augment)
ADF_residual = np.vstack(ADF_residual)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(4, 3))

offset = (np.arange(100)+1)*0.025
offset = np.min([offset, np.ones(100)*0.25], axis=0)

ax.set_ylabel("Validation Loss (MSE)")
ax.set_xlabel("Epochs")

category_colors = plt.colormaps['berlin'](np.linspace(0.15, 0.85, 3))
TSMixer_c = category_colors[0]
ADRAugment_c = category_colors[1]
ADRResidual_c = category_colors[2]

plt.errorbar(np.arange(100)+1-offset, np.mean(TSMixer, axis=0), yerr=np.std(TSMixer, axis=0), label='TSMixer', linestyle='-', capsize=1, color=TSMixer_c)
plt.errorbar(np.arange(100)+1, np.mean(ADF_augment, axis=0), yerr=np.std(ADF_augment, axis=0), label='TSMixer-ADRAugment', linestyle='-.', capsize=1, color=ADRAugment_c)
plt.errorbar(np.arange(100)+1+offset, np.mean(ADF_residual, axis=0), yerr=np.std(ADF_residual, axis=0), label='TSMixer-ADRResidual', linestyle="--", capsize=1, color=ADRResidual_c)
plt.legend()

ax.set_xscale('log')

plt.ticklabel_format(style='sci', axis='y', scilimits=(0,0))

plt.title("Validation Loss During Training (10 runs)")

plt.savefig("validation_loss_plot.pdf", bbox_inches='tight')
```
